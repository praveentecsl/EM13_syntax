```python
import pandas as pd

train_df = pd.read_csv('/content/train.csv')
test_df = pd.read_csv('/content/test.csv')

print("First 5 rows of train_df:")
display(train_df.head())

print("\nFirst 5 rows of test_df:")
display(test_df.head())
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Display the shape of the train_df DataFrame
print("Shape of train_df:")
print(train_df.shape)

# 2. Print the data types of each column in train_df
print("\nData types of train_df columns:")
print(train_df.dtypes)

# 3. Check for and display the number of missing values in each column of train_df
print("\nMissing values in train_df:")
print(train_df.isnull().sum())

# 4. Generate descriptive statistics for the numerical columns in train_df
print("\nDescriptive statistics of train_df:")
display(train_df.describe())

# 5. Investigate the distribution of the target variable target_price
plt.figure(figsize=(10, 6))
sns.histplot(train_df['target_price'], kde=True)
plt.title('Distribution of Target Price')
plt.xlabel('Target Price')
plt.ylabel('Frequency')
plt.show()

# 6. Examine the correlation matrix of the numerical features
plt.figure(figsize=(12, 10))
correlation_matrix = train_df.corr(numeric_only=True)
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', fmt=".2f")
plt.title('Correlation Matrix of Numerical Features')
plt.show()
```

## Preprocess the data


```python
# Handle missing values
for col in ['num_bathrooms', 'view_rating', 'basement_area']:
    median_val = train_df[col].median()
    train_df[col] = train_df[col].fillna(median_val)
    test_df[col] = test_df[col].fillna(median_val)

train_df['renovated_year'] = train_df['renovated_year'].fillna(0)
test_df['renovated_year'] = test_df['renovated_year'].fillna(0)

# Extract year and month from sale_date
train_df['sale_date'] = pd.to_datetime(train_df['sale_date'])
test_df['sale_date'] = pd.to_datetime(test_df['sale_date'])

train_df['sale_year'] = train_df['sale_date'].dt.year
train_df['sale_month'] = train_df['sale_date'].dt.month

test_df['sale_year'] = test_df['sale_date'].dt.year
test_df['sale_month'] = test_df['sale_date'].dt.month

# Drop original date and id columns
train_df = train_df.drop(['sale_date', 'house_id'], axis=1)
test_df = test_df.drop(['sale_date', 'house_id'], axis=1)

print("First 5 rows of train_df after handling missing values and feature engineering:")
display(train_df.head())

print("\nFirst 5 rows of test_df after handling missing values and feature engineering:")
display(test_df.head())
```

```python
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline

# Separate target variable
X_train = train_df.drop('target_price', axis=1)
y_train = train_df['target_price']
X_test = test_df.copy()

# Identify categorical and numerical columns (excluding the target)
categorical_features = ['zip_area', 'condition_index', 'construction_grade']
numerical_features = X_train.select_dtypes(include=['int64', 'float64']).columns.tolist()
numerical_features = [col for col in numerical_features if col not in categorical_features]

# Create preprocessing pipelines for numerical and categorical features
numerical_transformer = StandardScaler()
categorical_transformer = OneHotEncoder(handle_unknown='ignore')

# Create a column transformer to apply different transformations to different columns
preprocessor = ColumnTransformer(
    transformers=[
        ('num', numerical_transformer, numerical_features),
        ('cat', categorical_transformer, categorical_features)])

# Fit and transform the training data
X_train_processed = preprocessor.fit_transform(X_train)

# Transform the test data
X_test_processed = preprocessor.transform(X_test)

print("Shape of processed training features:", X_train_processed.shape)
print("Shape of processed test features:", X_test_processed.shape)
```

```python
# Inspect data types after initial cleaning
print("Data types of X_train before processing:")
print(X_train.dtypes)

print("\nData types of X_test before processing:")
print(X_test.dtypes)

# Convert categorical columns to object type to ensure OneHotEncoder handles them correctly
for col in categorical_features:
    X_train[col] = X_train[col].astype(str)
    X_test[col] = X_test[col].astype(str)

# Re-run the preprocessing steps
preprocessor = ColumnTransformer(
    transformers=[
        ('num', numerical_transformer, numerical_features),
        ('cat', categorical_transformer, categorical_features)])

# Fit and transform the training data
X_train_processed = preprocessor.fit_transform(X_train)

# Transform the test data
X_test_processed = preprocessor.transform(X_test)

print("Shape of processed training features:", X_train_processed.shape)
print("Shape of processed test features:", X_test_processed.shape)
```

```python
from sklearn.linear_model import LinearRegression

# Instantiate the Linear Regression model
linear_reg_model = LinearRegression()

# Train the model
linear_reg_model.fit(X_train_processed, y_train)

print("Linear Regression model trained successfully.")
```

## Evaluate the model

```python
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score
import numpy as np

# Make predictions on the training data
y_train_pred = linear_reg_model.predict(X_train_processed)

# Calculate evaluation metrics
rmse = np.sqrt(mean_squared_error(y_train, y_train_pred))
mae = mean_absolute_error(y_train, y_train_pred)
r2 = r2_score(y_train, y_train_pred)

# Print the metrics
print(f"Training RMSE: {rmse:.2f}")
print(f"Training MAE: {mae:.2f}")
print(f"Training R²: {r2:.2f}")
```
