```python
import pandas as pd

train_df = pd.read_csv('Heart_Attack_training_dataset.csv')
eval_df = pd.read_csv('Hear_Attack_evaluation_dataset.csv')

print("Training Data Head:")
display(train_df.head())

print("\nEvaluation Data Head:")
display(eval_df.head())

print("\nTraining Data Info:")
train_df.info()

print("\nEvaluation Data Info:")
eval_df.info()

print("\nTraining Data Description:")
display(train_df.describe())

print("\nEvaluation Data Description:")
display(eval_df.describe())

print("\nTraining Data Missing Values:")
print(train_df.isnull().sum())

print("\nEvaluation Data Missing Values:")
print(eval_df.isnull().sum())
```

## Data preprocessing



```python
from sklearn.preprocessing import StandardScaler
import numpy as np

# Handle 'bp' column separately - split into systolic and diastolic
def split_bp(df):
    bp_split = df['bp'].str.split('/', expand=True)
    df['systolic_bp'] = pd.to_numeric(bp_split[0])
    df['diastolic_bp'] = pd.to_numeric(bp_split[1])
    df = df.drop('bp', axis=1)
    return df

train_df = split_bp(train_df)
eval_df = split_bp(eval_df)

# Identify categorical columns (excluding 'patient_id' and the target)
categorical_cols = train_df.select_dtypes(include=['object']).columns.tolist()
if 'heart_attack_risk' in categorical_cols:
    categorical_cols.remove('heart_attack_risk')
if 'patient_id' in categorical_cols:
    categorical_cols.remove('patient_id')

eval_categorical_cols = eval_df.select_dtypes(include=['object']).columns.tolist()
if 'patient_id' in eval_categorical_cols:
    eval_categorical_cols.remove('patient_id')


# Apply one-hot encoding
train_df_encoded = pd.get_dummies(train_df, columns=categorical_cols, drop_first=True)
eval_df_encoded = pd.get_dummies(eval_df, columns=eval_categorical_cols, drop_first=True)

# Separate target variable
X_train = train_df_encoded.drop(['heart_attack_risk', 'patient_id'], axis=1)
y_train = train_df_encoded['heart_attack_risk']
eval_patient_ids = eval_df_encoded['patient_id']
eval_df_encoded = eval_df_encoded.drop('patient_id', axis=1)


# Identify numerical columns in the training data after encoding
numerical_cols = X_train.select_dtypes(include=np.number).columns.tolist()

# Apply standard scaling to numerical columns
scaler = StandardScaler()
X_train[numerical_cols] = scaler.fit_transform(X_train[numerical_cols])
eval_df_encoded[numerical_cols] = scaler.transform(eval_df_encoded[numerical_cols])

# Align columns - important for consistent feature sets
train_cols = X_train.columns
eval_cols = eval_df_encoded.columns

# Add missing columns to eval_df_encoded with default value 0
missing_in_eval = set(train_cols) - set(eval_cols)
for c in missing_in_eval:
    eval_df_encoded[c] = 0

# Add missing columns to X_train with default value 0
missing_in_train = set(eval_cols) - set(train_cols)
for c in missing_in_train:
    X_train[c] = 0

# Ensure the order of columns is the same
eval_df_encoded = eval_df_encoded[train_cols]

print("Processed Training Data Head:")
display(X_train.head())
print("\nProcessed Evaluation Data Head:")
display(eval_df_encoded.head())
print("\nTarget Variable Head:")
display(y_train.head())
```

## Feature Selection and Engineering
